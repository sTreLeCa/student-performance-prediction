```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler

df = pd.read_csv("../data/raw/student-mat.csv", sep=";")
# This file doesnt exist right now due to its large size. The raw file can be inserted if you wish but for file compression purposes we decided not to include it.

df.head()
```

```python
print("Shape:", df.shape)
df.info()
df.isnull().sum()
df.duplicated().sum()

df = df.drop_duplicates()
```

```python
target = "G3"

X = df.drop(columns=[target])
y = df[target]

categorical_cols = X.select_dtypes(include="object").columns
numerical_cols = X.select_dtypes(exclude="object").columns

X_encoded = pd.get_dummies(X, columns=categorical_cols, drop_first=True)
```

```python
scaler = StandardScaler()
X_encoded[numerical_cols] = scaler.fit_transform(
    X_encoded[numerical_cols]
)
```

```python
df_cleaned = pd.concat([X_encoded, y], axis=1)

df_cleaned.to_csv(
    "../data/processed/student_cleaned.csv",
    index=False
)

df_cleaned.head()
```
